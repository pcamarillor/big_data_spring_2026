```python
#Se crea el topic en kafka si este no existe
kafka-topics.sh --create \
  --topic videogames \
  --bootstrap-server localhost:9092
```

```python
from kafka import KafkaProducer
from faker import Faker
import json
import random
import time
from datetime import datetime

# Inicialización de Faker para datos sintéticos
fake = Faker()

# Configuración del Productor
producer = KafkaProducer(
    bootstrap_servers='kafka:9093',
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

GENRES = ["RPG", "Action", "FPS", "Sports", "Racing", "Survival Horror", "Simulator"]
PLATFORMS = ["PC", "PS5", "Xbox", "Switch"]

def generate():
    """Genera un diccionario simulando la compra de un videojuego."""
    return {
        "id": fake.uuid4(),
        "game": fake.word().title(),
        "genre": random.choice(GENRES),
        "platform": random.choice(PLATFORMS),
        "price": round(random.uniform(5, 80), 2),
        "timestamp": datetime.utcnow().isoformat() # Dato crucial para el TimestampType de PySpark
    }

print("Producing streaming data to 'videogames' topic...")

# Bucle infinito para simular el streaming en tiempo real
while True:
    try:
        data = generate()
        # Envío de datos al tópico "videogames"
        producer.send("videogames", data)
        print(f"Sent: {data}")
        time.sleep(1) # Pausa de 1 segundo entre mensajes
    except Exception as e:
        print(f"Error enviando datos: {e}")
        time.sleep(2) # Pausa un poco más larga si hay un error antes de reintentar
```
