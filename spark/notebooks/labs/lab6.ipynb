```python
from pcamarillor.spark_utils import SparkUtils

spark = SparkSession.builder \
    .appName("VideoGames Graph Analysis") \
    .getOrCreate()

# Leer dataset
df = spark.read \
    .option("header", "false") \
    .option("inferSchema", "true") \
    .csv("rec-amz-Video-Games.edges")

# Renombrar columnas
df = df.toDF("user", "product", "rating", "timestamp")

df.show(5)
df.printSchema()
```
